### How to Approach any Analytics Problem 🤷??
***This is a generic systematic approach which can be tailor-made to as per requirement***

#### 1. Define the Problem Clearly 📝

- What is the objective?
    - Understand the business context or the goal behind the analysis.
    - Example: “What drives sales for product X?” or “How can an investor optimize their portfolio?”
- Who are the stakeholders?
    - Identify who will use the insights and tailor your analysis accordingly.

#### 1.5 Generate Hypotheses 🧠

- What could be driving the outcomes?
    - Think critically and creatively about potential reasons for observed behaviors or trends.
    - Leverage Domain Knowledge:
- Collaborate with stakeholders or SMEs (subject matter experts) to ensure hypotheses align with business realities.
- Formulate Testable Statements:
    - Example: "Gold prices increase during economic instability," or, "Quarterly volatility of silver is higher than gold."

#### 2. Understand the Data 💾

- What data is available?
    - Check the type, source, and format of data. Examples: structured (tables, CSVs) or unstructured (text, images).
    - What is missing?
- Identify any gaps in the data and consider sourcing additional information if necessary.

#### 3. Formulate Analytics Questions 🧪

- Break the main problem into smaller questions, such as:
    - Descriptive: What has happened so far?
    - Diagnostic: Why did it happen?
    - Predictive: What is likely to happen next?
    - Prescriptive: What actions should be taken?

#### 4. Explore and Preprocess the Data 🔍

- EDA (Exploratory Data Analysis):
    - Identify trends, correlations, and anomalies. Use tools like pandas and matplotlib for visualization.
- Data Cleaning:
    - Handle missing values, remove duplicates, and fix inconsistencies.
- Data Transformation:
    - Normalize, scale, and encode the data, depending on the requirements.

#### 4.5. Test the Hypotheses 🔬

- Validate hypotheses with data:
    - Use statistical techniques (e.g., correlation analysis, regression) or visualization.
    - Example: Plot inflation vs. gold prices to check correlation visually and calculate statistical significance.
- Accept or Reject:
    - If a hypothesis is rejected, refine it or explore alternative explanations.

#### 5. Select the Right Analytics Technique ✅

- Match the question to the method:
    - Trend analysis → Descriptive analytics
    - Root cause identification → Diagnostic analytics
    - Forecasting → Predictive analytics
    - Optimization → Prescriptive analytics
- Examples of techniques:
    - Statistical analysis, machine learning, hypothesis testing, or data visualization.

#### 6. Conduct the Analysis 🧮

- Apply the chosen techniques and document each step:
    - Descriptive Statistics: Summarize mean, median, mode, etc.
    - Modeling: Train machine learning models or fit statistical models.
    - Visualization: Use charts to simplify insights for stakeholders.
    - There are other techniques which can be statistical, optimization, etc. 

#### 7. Validate Insights and Results 📐

- Are results accurate?
    - Test models on unseen data to evaluate accuracy.
- Cross-check findings:
    - Ensure alignment with domain knowledge and business context.

#### 8. Communicate Results Effectively 🎙️

- Simplify technical findings for stakeholders:
    - Visuals: Use dashboards, charts, and infographics.
    - Stories: Relate findings to business goals or user stories.
- Be ready to answer questions and provide actionable recommendations.

#### 9. Suggest Actions and Solutions 🛣️

- Translate analytics into business impact:
    - Example: "Invest 60% in gold and 40% in silver for lower risk, considering past trends."
    -  Example: "Implement predictive maintenance to reduce costs by X%."


#### 10. Iterate and Monitor ♾️🚨🕵🏻

- Analytics is iterative. Keep improving your models and refine insights as new data becomes available.
- Set up feedback loops to assess the effectiveness of implemented actions.


#### Tips for a Smooth Approach 📌

1. Start Simple: Focus on key questions first and build complexity step-by-step.
2. Be Visual: Use graphs to uncover trends and engage stakeholders.
3. Involve Stakeholders: Get regular feedback to ensure alignment with goals.
4. Keep Learning: Evolve your skills with new tools and techniques as analytics advances.

### Step 1: Calculate Returns 🔢
Instead of raw returns, we can use **logarithmic returns**

https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.shift.html

```python
# Compute returns without pct_change()
returns = ((data - data.shift(1)) / data.shift(1)).dropna() # Alternate for pct_change()

# Compute log returns = ln(previous day/current day)
log_returns = np.log(data / data.shift(1)).dropna()

# Display first few returns
print('Simple Returns')
display(returns.head())
print('Log Returns')
display(log_returns.head())
```

## Step 4: Monte Carlo GBM Simulation 🎲
### The GBM formula:
$$
\ [
    S_{t} = S_{t-1} \times e^{(\mu - 0.5 \sigma^2) \Delta t \space + \space \sigma \sqrt{\Delta t} Z}
\ ]
\newline
\ [
    S_{t} = S_{t-1} \times e^{drift\space + \space diffusion}
\ ]

$$
Where:
$$ \ ( S_{t-1} \ ) = Previous \space price $$
$$ \ ( \mu \ ) = Expected \space return $$
$$ \ ( \sigma \ ) = Volatility $$
$$ \ ( \Delta t \ ) = Time \space step (1 day) $$
$$ \ ( Z \ ) = Random \space normal \space variable $$

### Key Additions:
- **Cholesky Decomposition** to generate correlated random values.
- Cholesky decomposition is a mathematical technique that allows you to break down a correlation matrix (representing the relationships between multiple variables) into a lower triangular matrix and its conjugate transpose
- It enables the generation of correlated random variables by transforming a set of uncorrelated variables using this matrix; 
- Essentially, it lets you "build" correlated data from a set of independent variables based on the specified correlation structure. 

```python
# Simulation parameters
T, dt, n_sim = 3 * 252, 1, 10000  # 3 years, 252 trading days/year, 10000 simulations
S0 = data.iloc[-1].values  # Last known prices
L = np.linalg.cholesky(correlation_matrix.values)  # Cholesky decomposition for correlation

# Initialize paths
paths = np.zeros((T, 2, n_sim))
paths[0, :, :] = S0.reshape(-1, 1)

# Simulate correlated GBM paths
np.random.seed(42)
for t in range(1, T):
    Z = np.random.randn(2, n_sim)  # Independent normal variables
    Z_corr = L @ Z  # Apply correlation
    drift = (mu - 0.5 * sigma**2) * dt
    diffusion = sigma.values[:, None] * np.sqrt(dt) * Z_corr
    paths[t] = paths[t - 1] * np.exp(drift.values[:, None] + diffusion)

print("Monte Carlo Simulation Completed! ✅")
```

## Step 6: Portfolio Simulation 💰
We assume:
- **70% allocation in gold**
- **30% allocation in silver**

The total portfolio value is:
\[
Portfolio_t = 0.7 \times Gold_t + 0.3 \times Silver_t
\]


```python
# Compute portfolio paths
portfolio = 0.7 * paths[:, 0, :] + 0.3 * paths[:, 1, :]

# Plot simulated portfolio paths
plt.figure(figsize=(10,5))
plt.plot(portfolio[:, :10], alpha=0.6)
plt.title("Simulated Portfolio Value (70% Gold, 30% Silver)")
plt.xlabel("Days")
plt.ylabel("Portfolio Value")
plt.show()
```
